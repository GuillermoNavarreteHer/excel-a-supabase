# Crear un Excel y subirlo a Supabase
Cambia `TU_PROYECTO` y `TU_API_KEY` y ejecuta las celdas en orden.

```
!pip install pandas openpyxl supabase -q
```

```
SUPABASE_URL = "https://TU_PROYECTO.supabase.co"
SUPABASE_KEY = "TU_API_KEY"
BUCKET = "archivos"
ARCHIVO = "datos.xlsx"
```

```
import pandas as pd

df = pd.DataFrame({"nombre": ["Ana", "Luis"], "edad": [25, 30]})
df.to_excel(ARCHIVO, index=False)
df
```

```
from supabase import create_client

sb = create_client(SUPABASE_URL, SUPABASE_KEY)
with open(ARCHIVO, "rb") as f:
    sb.storage.from_(BUCKET).upload(ARCHIVO, f.read(), {
        "content-type": "application/vnd.openxmlformats-officedocument.spreadsheetml.sheet",
        "upsert": "true",
    })
print("Subido!")
```

## Opcion: guardar los datos en una tabla
La tabla `personas` debe existir con las columnas `nombre` y `edad`.

```
sb.table("personas").insert(df.to_dict("records")).execute()
```
